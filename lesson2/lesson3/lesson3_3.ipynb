```python
import random

min = 1
max = 100
count = 0
target = random.randint(min,max)
print("=====猜數字遊戲=====\n")
while True:
    keyin = int(input(f"猜數字範圍{min}到{max}之間的數字: "))
    count += 1
    if keyin >= min and keyin <= max:
        if target == keyin:
            print(f"猜對了，答案是:{target}")
            break
        elif (keyin > target):
            print("太大，再小一點")
            max = keyin - 1
        elif (keyin < target):
            print("太小，再大一點")
            min = keyin + 1
    else:
        print("請輸入提示範圍內的數字")

print("遊戲結束")
```
